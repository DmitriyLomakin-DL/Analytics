```python
import pandas as pd

df = pd.DataFrame({
    "product": ["avia", "train", "hotel"],
    "revenue": [150000, 90000, 70000]
})

df
```

```python
import duckdb

duckdb.sql("""
    SELECT
        product,
        revenue
    FROM df
    ORDER BY revenue DESC
""")
```

```python
duckdb.sql("""
    SELECT
        product,
        revenue
    FROM df
    WHERE revenue > 80000
""").df()
```

```python
import matplotlib.pyplot as plt

plt.bar(df["product"], df["revenue"])
plt.title("Revenue by product")
plt.xlabel("Product")
plt.ylabel("Revenue")
plt.show()
```
